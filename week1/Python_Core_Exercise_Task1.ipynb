```python
#Task 1.1
import random
from datetime import datetime, timedelta

DEVICE_TYPES = [
    "Temperature_Sensor",
    "Humidity_Sensor",
    "Pressure_Sensor",
    "Light_Sensor",
    "Motion_Sensor"
]

LOCATIONS = [
    "Room A",
    "Room B",
    "Lab 1",
    "Lab 2",
    "Entrance"
]

READING_RANGES = {
    "Temperature_Sensor": (15.0, 40.0),
    "Humidity_Sensor": (20.0, 90.0),
    "Pressure_Sensor": (950.0, 1050.0),
    "Light_Sensor": (0.0, 1000.0),
    "Motion_Sensor": (0.0, 1.0)
}

raw_sensor_data = []

start_time = datetime.now()

for i in range(1, 101):
    timestamp = start_time + timedelta(seconds=i)
    device_id = f"DEV{i:03d}"
    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)

    min_value, max_value = READING_RANGES[device_type]
    value = round(random.uniform(min_value, max_value), 2)

    raw_sensor_data.append((
        timestamp.strftime("%Y-%m-%d %H:%M:%S"),
        device_id,
        device_type,
        location,
        value
    ))

print(raw_sensor_data)
```

```python
#Task 1.2
temperature_readings = []

for timestamp, device_id, device_type, location, value in raw_sensor_data:
    if device_type == "Temperature_Sensor":
        temperature_readings.append({
            "timestamp": timestamp,
            "device_id": device_id,
            "location": location,
            "temperature_celsius": value
        })

print(temperature_readings)
```

```python
#Task 1.3
unique_locations = set()

for timestamp, device_id, device_type, location, value in raw_sensor_data:
    unique_locations.add(location)

print(unique_locations)
```

```python
#Task 1.4
device_stats = {}

# Initialize min and max values
for device_type in DEVICE_TYPES:
    device_stats[device_type] = (float('inf'), float('-inf'))

# Find the minimum and maximum values
for timestamp, device_id, device_type, location, value in raw_sensor_data:
    min_value, max_value = device_stats[device_type]

    min_value = min(min_value, value)
    max_value = max(max_value, value)

    device_stats[device_type] = (min_value, max_value)

print(device_stats)
```
